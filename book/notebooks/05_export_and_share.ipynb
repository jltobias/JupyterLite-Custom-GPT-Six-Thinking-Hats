# 05 — Download and share your saved decision

This cell reopens `decision.json`, including any Purple synthesis saved in Notebook 06. It validates the record and prepares all formats before writing. Every export creates a new uniquely named snapshot folder under `exports/`. All four files belong to that snapshot; earlier snapshots remain unchanged. An incomplete new snapshot is removed if a write fails.

- **JSON:** complete editable data for reopening later.
- **SVG:** complete connected map, scalable and editable as text.
- **Markdown:** readable portable text.
- **HTML:** standalone semantic headings and lists, plus an expandable map; no network service is required to read it.

Review the contents and source context before sharing.

First save a decision in **02**, then use **Run → Run All Cells** here. **Action needed** means no successful export is being reported; fix the indicated input or storage problem and rerun.


```python
from IPython.display import HTML, display
from decision_workflow import load_record, export_record, download_links, ValidationError

record = None
paths = None
try:
    record = load_record('decision.json')
    paths = export_record(record, 'exports')
except (ValidationError, OSError) as error:
    record = None
    paths = None
    print('Action needed: export did not finish.', error)
    print('Check that Notebook 02 saved a valid decision.json and that browser storage is available, then rerun this cell.')
else:
    print('Exported saved decision:', record['title'])
    print('Snapshot folder:', paths['json'].parent)
    try:
        display(HTML(download_links(paths)))
    except OSError as error:
        print('Files saved, but embedded download links could not be prepared:', error)
        print('Use the file browser to open the snapshot folder above, right-click each file, and choose Download.')
```
