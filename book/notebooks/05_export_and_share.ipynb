# 05 — Export and share

Export both the source record and the visualization. JSON preserves editable structure; SVG preserves a scalable visual; HTML gives a simple standalone artifact.

```python
import json
from pathlib import Path
from IPython.display import FileLink, display
from six_hats_mindmap import render_mind_map
record={'title':'Export demo','question':'How should this artifact be shared?','white':{'items':['JSON is machine-readable']},'red':{'items':['The map should feel easy to revisit']},'black':{'items':['Do not publish sensitive content']},'yellow':{'items':['Portable files make review easier']},'green':{'items':['Bundle provenance with the artifact']},'blue':{'items':['Review, export, then publish intentionally']}}
out=Path('exports'); out.mkdir(exist_ok=True)
svg=render_mind_map(record)
(out/'six-hats.json').write_text(json.dumps(record,indent=2))
(out/'six-hats.svg').write_text(svg)
(out/'six-hats.html').write_text("<!doctype html><meta charset='utf-8'>"+svg)
for name in ['six-hats.json','six-hats.svg','six-hats.html']: display(FileLink(str(out/name)))
```
