# 03 — Reopen and visualize your decision

The original **Six Thinking Hats** method is Edward de Bono's. Purple is an optional project extension. Keep these notebooks in the same folder so they share `decision.json`.

Run Notebook 02 first. This notebook reads your saved record and retains every item, key question, evidence annotation, optional Purple conclusion, and possible outcome. The connected SVG grows with its contents. Use the semantic HTML view below on smaller screens.

This notebook needs a saved `decision.json` from **02**. If you see **Action needed**, follow that message and rerun the cell; it has not produced a new map. Both view cells reopen the saved file independently, including after a kernel restart.


```python
from IPython.display import SVG, display
from decision_workflow import load_record, ValidationError
from six_hats_mindmap import render_mind_map

record = None
try:
    record = load_record('decision.json')
    print('Reopened:', record['title'])
    display(SVG(render_mind_map(record)))
except (ValidationError, OSError) as error:
    record = None
    print('Action needed:', error)
    print('Save or repair decision.json in Notebook 02, then rerun this cell. For a ready-to-run example, open Notebook 04.')
```

```python
from IPython.display import HTML, display
from decision_workflow import load_record, to_html, ValidationError

# Reopen the file so this cell also works after a kernel restart.
record = None
try:
    record = load_record('decision.json')
    display(HTML(to_html(record)))
except (ValidationError, OSError) as error:
    record = None
    print('Action needed:', error)
    print('Save or repair decision.json in Notebook 02, then rerun this cell.')
```
