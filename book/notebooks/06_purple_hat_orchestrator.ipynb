# 06 — Purple Hat orchestrator (project extension)

The original Six Thinking Hats method uses six hats. This notebook adds an optional **Purple Hat — Orchestrator** for this repository. It is a practical synthesis layer, not a seventh de Bono thinking mode.

```python
def purple_orchestrator(data):
    white=data.get('white',{}).get('items',[])
    black=data.get('black',{}).get('items',[])
    yellow=data.get('yellow',{}).get('items',[])
    green=data.get('green',{}).get('items',[])
    blue=data.get('blue',{}).get('items',[])
    return {'label':'Purple Hat — Orchestrator','items':[f'Evidence base contains {len(white)} white-hat items; check which are sourced.',f'Balance {len(yellow)} opportunity items against {len(black)} risk items.',f'Prioritize among {len(green)} creative alternatives rather than pursuing all of them.','Identify missing stakeholder, equity, privacy, or feasibility perspectives.','Convert the blue-hat process into owners, dates, and success measures.' if blue else 'Add a blue-hat action plan.']}
sample={'white':{'items':['Evidence A','Unknown B']},'black':{'items':['Risk A']},'yellow':{'items':['Benefit A']},'green':{'items':['Experiment A','Experiment B']},'blue':{'items':['Define metrics']}}
sample['purple']=purple_orchestrator(sample)
sample['purple']
```
