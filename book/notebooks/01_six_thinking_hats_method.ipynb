## What a record contains

Required: non-empty `title` and `question`, plus six hat objects. Each hat requires an `items` list of strings. Optional `label`, `focus`, and `questions` preserve the headings and key questions. Empty lists mean no entries have been recorded.

Optional record fields: `subtitle`, `context`, `demo_notice`, `evidence`, `outcomes`, and `purple`. Evidence entries contain `claim`, `source`, `status`, and optional `notes` strings. A URL alone does not establish a claim. `outcomes` is a list of possible results, not achieved results.

Purple can add `author`, `recommendation`, `themes`, `tensions`, `blind_spots`, `uncertainties`, `decision_pivots`, and `next_steps`. All but author and recommendation are lists of strings. See the workflow guide for the complete schema.
