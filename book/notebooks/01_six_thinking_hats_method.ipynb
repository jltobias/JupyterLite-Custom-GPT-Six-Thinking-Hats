# 01 — Six Thinking Hats method and schema

This notebook turns Edward de Bono's **Six Thinking Hats** method into a machine-readable schema for JupyterLite.

**Citation:** de Bono, E. (1985). *Six Thinking Hats*. Little, Brown. Goodreads: https://www.goodreads.com/book/show/97030.Six_Thinking_Hats

The original six hats are white, red, black, yellow, green, and blue. The Purple Hat used elsewhere in this repository is a project-specific extension.

```python
hats = {
    'white': 'facts, evidence, knowns, unknowns',
    'red': 'feelings, intuition, values, reactions',
    'black': 'risks, cautions, constraints, failure modes',
    'yellow': 'benefits, value, opportunities, reasons it may work',
    'green': 'creative alternatives, experiments, new ideas',
    'blue': 'process, sequence, decisions, next steps',
}
for name, focus in hats.items():
    print(f'{name.title():6} -> {focus}')
```

```python
record = {'title':'Decision title','subtitle':'Optional context','question':'What are we trying to answer?'}
for name, focus in hats.items():
    record[name] = {'label':f'{name.title()} Hat','focus':focus,'items':[],'questions':[]}
record
```
