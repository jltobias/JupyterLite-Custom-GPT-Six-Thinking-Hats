```python
CENTRAL_QUESTION = 'How can I evaluate this decision using six complementary perspectives?'
CONTEXT = 'Replace this with background, constraints, stakeholders, and evidence links.'
PROMPT = f'''Use Edward de Bono's Six Thinking Hats method to examine the question below.
Return ONLY valid JSON.
Question: {CENTRAL_QUESTION}
Context: {CONTEXT}
Required keys: title, subtitle, question, white, red, black, yellow, green, blue.
For each hat return label, focus, 3-6 concise items, and 2-4 concise questions.
Do not invent citations.'''
print(PROMPT)
```

```python
import json
GPT_RESPONSE = r'''{"title":"Example","subtitle":"Replace with your GPT result","question":"How should we proceed?","white":{"items":["Known fact","Important unknown"]},"red":{"items":["Excitement"]},"black":{"items":["Plausible failure mode"]},"yellow":{"items":["Meaningful upside"]},"green":{"items":["Try a small experiment"]},"blue":{"items":["Define a success metric","Run a time-boxed pilot"]}}'''
data = json.loads(GPT_RESPONSE)
required={'title','question','white','red','black','yellow','green','blue'}
assert not (required-data.keys())
print('Validated:', data['title'])
```
