````python
CENTRAL_QUESTION = 'What decision should we examine?'
CONTEXT = 'Describe the stakeholders, constraints, timescale, and evidence.'
GPT_RESPONSE = r'''
'''
USE_DEMO = False
REPLACE_SAVED = False  # Set True only to deliberately replace an existing record with your pasted JSON.
````

````python
from decision_workflow import capture_prompt, ValidationError

if not all(name in globals() for name in ('CENTRAL_QUESTION', 'CONTEXT')):
    print('Action needed: run the input cell above first, then rerun this prompt cell.')
else:
    try:
        print(capture_prompt(CENTRAL_QUESTION, CONTEXT))
    except ValidationError as error:
        print('Action needed:', error)
        print('Correct the question/context, run the input cell, then rerun this cell.')
````

````python
from pathlib import Path
from decision_workflow import capture_response, load_record, save_record, ValidationError

RECORD_PATH = Path('decision.json')
data = None  # A failed attempt must not retain a previous record in this kernel.
if not all(name in globals() for name in ('GPT_RESPONSE', 'USE_DEMO', 'REPLACE_SAVED')):
    print('Action needed: run the input cell above first, then rerun this save cell.')
else:
    try:
        if type(USE_DEMO) is not bool or type(REPLACE_SAVED) is not bool:
            raise ValidationError('USE_DEMO and REPLACE_SAVED must be True or False without quotes.')
        if not isinstance(GPT_RESPONSE, str):
            raise ValidationError('GPT_RESPONSE must contain JSON text. Correct the quoted input or uploaded file and rerun the input cell.')
        if GPT_RESPONSE.strip():
            data = capture_response(GPT_RESPONSE, RECORD_PATH, overwrite=REPLACE_SAVED)
            print('Saved your decision:', data['title'])
        elif USE_DEMO and not RECORD_PATH.exists():
            data = load_record('examples/global_health.json')
            data.pop('purple', None)  # Begin practice with six hats; Notebook 06 adds synthesis.
            save_record(data, RECORD_PATH)
            print('ILLUSTRATIVE DEMO saved:', data['title'])
        elif RECORD_PATH.exists():
            data = load_record(RECORD_PATH)
            print('Reopened existing decision; no file changed:', data['title'])
        else:
            print('Action needed: no decision saved yet. Paste JSON or set USE_DEMO = True, run the input cell, then rerun this save cell. Continue to 03 only after a saved/reopened message.')
    except (ValidationError, OSError) as error:
        data = None
        print('Action needed:', error)
        print('Correct the input or file problem, run the input cell, then rerun this save cell. Existing decision content has not been replaced.')
````

## Upload JSON instead of pasting

If valid JSON contains three consecutive apostrophes (`'''`), it cannot go inside the default raw triple-apostrophe literal. Save it as UTF-8 `response.json` and upload it beside this notebook. In the input cell, replace the entire `GPT_RESPONSE = r...` assignment (including its closing quotes) with:

```python
from pathlib import Path
GPT_RESPONSE = None  # Clear an older response before attempting the read.
try:
    GPT_RESPONSE = Path("response.json").read_text(encoding="utf-8")
except (OSError, UnicodeError) as error:
    print("Action needed: could not read response.json.", error)
```

Keep the other input settings, execute the edited input cell, then execute the save cell. This alternative also works for long responses.
